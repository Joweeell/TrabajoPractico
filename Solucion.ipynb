1. Introduccion

**Poblacion:** Todos los estudiantes ingresados a la educacion superior en chile.

**Muestra:** 151.825 estudiantes pertenecientes a la region metrpolitana (Base de datos).

**Variables Cualitativas escogidas:**
- **GENERO:** Cualitativa Nominal.
- **TIPO DE GENERO:** Cualitativa Nominal.
- **AREA CONOCIMIENTO:** Cualitativa Nominal.

**Variables Cuantitativas escogidas:**
- **EDAD:** Cuantitativa Continua
- **DURACION TOTAL CARRERA (SEMESTRE):** Cuantitativa Discreta
- **VALOR ARANCEL (PESOS):** Cuantitativa Continua (variable continua requerida)

```python
import pandas as pd
import numpy as np

# Base de datos cargada
df = pd.read_excel('07_MATRICULAS_ED_SUPERIOR_METROPOLITANA_2021_ZONA NORTE.xlsx')
df
```

```python

print("Primeras 5 filas")
df.head(5)
```

```python
n = df.shape[0]
print("Tamaño de la muestra:", n)
```

**2. Item 1: Variables Cualitativas:**

**VARIABLE 1: GENERO (Cualitativa Nominal)**

```python

#Cálculo de frecuencia absoluta. La palabra frecuencia representa la frecuencia absoluta.
frecuencia = df.groupby('GENERO').size()

n = df.shape[0]

#Crear el DF con frecuencias absolutas y relativas
tabla = pd.DataFrame({
    'fi': frecuencia,
    'hi': round((frecuencia/n) *100, 1).astype(str) + '%'  # redondear según contexto y valores de los datos.
})
print("--- TABLA DE FRECUENCIA: GENERO ---")
tabla
```

**GRAFICO DE BARRAS PARA GENERO**

```python
import matplotlib.pyplot as plt

#Gráfico de barras

#Considerar que "tabla" es el nombre de la tabla de frecuencias construida

#Variables
categorias = tabla.index
f = tabla['fi']   #Considerar que "Frecuencia" es el tipo de frecuencia a graficar

#Gráfico
fig, ax = plt.subplots(figsize = (6,6))
barras = ax.bar(categorias, f, color = 'r', edgecolor ='black') #Los colores son de libre elección

#Personalización
ax.set_title('Distribucion de Estudiantes por Genero')
ax.set_xlabel('Genero')
ax.set_ylabel('Cantidad de estudiantes (fi)')
ax.bar_label(barras)
plt.xticks(rotation=45) #Rota en 45° las categorías para evitar la superposición
plt.show()
```

**INTERPRETACION DEL GRAFICO:** Observando el grafico muestra la cantidad de Estudiantes por Genero. Existe una mayor presencia de estudiantes de genero femenino, con un total de 80,617 registros, frente a 71,208 estudiantes de genero masculino.

**2. Item 1: Variables Cuantitativas**

**Variable 4: EDAD (Cuantitativa Continua)**

```python
#Cálculo de frecuencia absoluta. La palabra frecuencia representa la frecuencia absoluta.
frecuencia = df.groupby('EDAD').size()

n = df.shape[0]

#Crear el DF con todas las frecuencias
tabla = pd.DataFrame({
    'fi': frecuencia,
    'hi': round((frecuencia/n) *100, 1).astype(str) + '%', # redondear según contexto y valores de los datos, mostrando símbolo de '%'
    'Fi':frecuencia.cumsum(),
    'Hi': round((frecuencia/n) * 100,2).cumsum().astype(str) + '%'
})
print("--- TABLA DE FRECUENCIA: EDAD ---")
tabla
```

```python
#Gráfico de barras

#Considerar que "tabla" es el nombre de la tabla de frecuencias construida

#Variables
categorias = tabla.index
f = tabla['fi']   #Considerar que "Frecuencia" es el tipo de frecuencia a graficar

#Gráfico
fig, ax = plt.subplots(figsize = (30,6))
barras = ax.bar(categorias, f, color = 'r', edgecolor ='black') #Los colores son de libre elección

#Personalización
ax.set_title('Distribucion de Estudiantes por Edad')
ax.set_xlabel('Edad (Años)')
ax.set_ylabel('Cantidad de Alumnos (fi)')
ax.bar_label(barras)
plt.xticks(rotation=45) #Rota en 45° las categorías para evitar la superposición
plt.show()
```

**INTERPRETACION DEL GRAFICO:** El gráfico refleja la distribución de los alumnos según su edad. Se evidencia una clara concentración en el tramo de jóvenes de **19 a 24 años**, destacando los **19 años como la edad con mayor volumen** al alcanzar 16,392 estudiantes, seguida por los **24 años con 14,139 y los 21 años con 13,825 inscritos**. Por el contrario, a medida que la edad aumenta la matrícula disminuye progresivamente, registrándose **la menor cantidad en los estudiantes de 81 años**, con apenas 1 registro en total.

```python
# Medidas de tendencia central
media = round(df['EDAD'].mean(),1)
mediana = round(df['EDAD'].median(),1)
moda = df['EDAD'].mode()[0] # El [0] es para que en el print no quede -> Moda: 0    19

# Percentiles
#ingrese el valor "k" en decimal
percentil_25 = round(df['EDAD'].quantile(0.25),1) # 25% 
percentil_50 = round(df['EDAD'].quantile(0.50),1) # 50%
percentil_75 = round(df['EDAD'].quantile(0.75),1) # 75%

# Medidas de dispersión
rango = df['EDAD'].max() - df['EDAD'].min()
varianza = round(df['EDAD'].var(),1)
desviacion_estandar = round(df['EDAD'].std(),1)
coeficiente_variacion = round((desviacion_estandar / media)*100,1).astype(str) + '%'

print("--- MEDIDAS DE TENDENCIA CENTRAL ---")
print("Media:", media)
print("Mediana:", mediana)
print("Moda:", moda)

print("\n--- PERCENTILES ---")
print("Percentil 25%:", percentil_25)
print("Percentil 50%:", percentil_50)
print("Percentil 75%:", percentil_75)

print("\n--- MEDIDAS DE DISPERSIÓN ---")
print("Rango:", rango)
print("Varianza:", varianza)
print("Desviación Estándar:", desviacion_estandar)
print("Coeficiente de Variación:", coeficiente_variacion)
```

**INTERPRETACION DE MEDIDAS DE TENDENCIA CENTRAL:** El promedio de edad de los estudiantes matriculados es de 27 años, mientras que la edad más frecuente es de 19 años (Moda). Asimismo, el 50% de los alumnos tiene 24 años o menos (Mediana).

**INTERPRETACION DE PERCENTILES:** El 25% de los estudiantes más jóvenes tiene 21 años o menos, el 50% registra 24 años o menos, y el 75% de la matrícula tiene 32 años o menos (lo que implica que solo un 25% supera los 32 años).

**INTERPRETACION DE MEDIDAS DE DISPERSIÓN:** Existe una brecha de 63 años entre el estudiante de mayor y menor edad. Las edades se alejan en promedio 8.6 años respecto a la media, y al obtener un Coeficiente de Variación de 30.9%, se evidencia una alta heterogeneidad o variabilidad en la edad del alumnado.

**Variable 5: DURACION TOTAL CARRERA (SEMESTRE) (Cuantitativa Discreta)**

```python
#Cálculo de frecuencia absoluta. La palabra frecuencia representa la frecuencia absoluta.
frecuencia = df.groupby('DURACION TOTAL CARRERA (SEMESTRES)').size()

n = df.shape[0]

#Crear el DF con todas las frecuencias
tabla = pd.DataFrame({
    'fi': frecuencia,
    'hi': round((frecuencia/n) *100, 1).astype(str) + '%', # redondear según contexto y valores de los datos, mostrando símbolo de '%'
    'Fi':frecuencia.cumsum(),
    'Hi': ((frecuencia/n) * 100).cumsum()
})
tabla
```

```python
#Gráfico de barras

#Considerar que "tabla" es el nombre de la tabla de frecuencias construida

#Variables
categorias = tabla.index.astype(str)
f = tabla['fi']   #Considerar que "Frecuencia" es el tipo de frecuencia a graficar

#Gráfico
fig, ax = plt.subplots(figsize = (10,6))
barras = ax.bar(categorias, f, color = 'r', edgecolor ='black') #Los colores son de libre elección

#Personalización
ax.set_title('Distribución de Estudiantes según Duración Total de la Carrera')
ax.set_xlabel('Duración Total (Semestres)')
ax.set_ylabel('Cantidad de Alumnos')
ax.bar_label(barras)
plt.xticks(rotation=0) #Rota en 45° las categorías para evitar la superposición
plt.show()
```

**INTERPRETACION DEL GRAFICO:** El gráfico refleja la distribución de los alumnos según la duración total de su carrera. Se evidencia una clara concentración en las carreras de 5 a 10 semestres, destacando los 10 semestres (5 años) como la duración con mayor volumen al alcanzar 45.282 estudiantes, seguida por los 8 semestres (4 años) con 32.520 y los 5 semestres (2.5 años) con 28.840 inscritos. Por el contrario, las carreras con duraciones impares o excesivamente extensas presentan una presencia marginal, registrándose la menor cantidad en las carreras de 16 semestres, con apenas 4 estudiantes en total.

```python
# Medidas de tendencia central
media = round(df['DURACION TOTAL CARRERA (SEMESTRES)'].mean(),1)
mediana = round(df['DURACION TOTAL CARRERA (SEMESTRES)'].median(),1)
moda = df['DURACION TOTAL CARRERA (SEMESTRES)'].mode()[0] # El [0] es para que en el print no quede -> Moda: 0    19

# Percentiles
#ingrese el valor "k" en decimal
percentil_25 = round(df['DURACION TOTAL CARRERA (SEMESTRES)'].quantile(0.25),1) # 25% 
percentil_50 = round(df['DURACION TOTAL CARRERA (SEMESTRES)'].quantile(0.50),1) # 50%
percentil_75 = round(df['DURACION TOTAL CARRERA (SEMESTRES)'].quantile(0.75),1) # 75%

# Medidas de dispersión
rango = df['DURACION TOTAL CARRERA (SEMESTRES)'].max() - df['DURACION TOTAL CARRERA (SEMESTRES)'].min()
varianza = round(df['DURACION TOTAL CARRERA (SEMESTRES)'].var(),1)
desviacion_estandar = round(df['DURACION TOTAL CARRERA (SEMESTRES)'].std(),1)
coeficiente_variacion = round((desviacion_estandar / media)*100,1).astype(str) + '%'

print("--- MEDIDAS DE TENDENCIA CENTRAL ---")
print("Media:", media)
print("Mediana:", mediana)
print("Moda:", moda)

print("\n--- PERCENTILES ---")
print("Percentil 25%:", percentil_25)
print("Percentil 50%:", percentil_50)
print("Percentil 75%:", percentil_75)

print("\n--- MEDIDAS DE DISPERSIÓN ---")
print("Rango:", rango)
print("Varianza:", varianza)
print("Desviación Estándar:", desviacion_estandar)
print("Coeficiente de Variación:", coeficiente_variacion)
```

**INTERPRETACION DE MEDIDAS DE TENDENCIA CENTRAL:** El promedio de la duración total de las carreras es de 7.6 semestres, mientras que la duración más frecuente es de 10 semestres (Moda). Asimismo, el 50% de los alumnos cursa carreras de 8.0 semestres o menos (Mediana).

**INTERPRETACION DE PERCENTILES:** El 25% de los estudiantes cursa carreras de 5.0 semestres o menos, el 50% registra 8.0 semestres o menos, y el 75% de la matrícula estudia carreras de 10.0 semestres o menos (lo que implica que solo un 25% supera los 10.0 semestres de duración).

**INTERPRETACION DE MEDIDAS DE DISPERSIÓN:** Existe una brecha de 23 semestres entre la carrera de mayor y menor duración. La duración de los programas se aleja en promedio 2.9 semestres respecto a la media, y al obtener un Coeficiente de Variación de 38.2%, se evidencia una alta heterogeneidad o variabilidad en la duración total de las carreras.

**Variable 6: VALOR ARANCEL (PESOS) (Cuantitativa Continua)**
