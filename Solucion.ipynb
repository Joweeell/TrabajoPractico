1. Introduccion

**Poblacion:** Todos los estudiantes ingresados a la educacion superior en chile.

**Muestra:** 151.825 estudiantes pertenecientes a la region metrpolitana (Base de datos).

**Variables Cualitativas escogidas:**
- **GENERO:** Cualitativa Nominal.
- **TIPO DE GENERO:** Cualitativa Nominal.
- **AREA CONOCIMIENTO:** Cualitativa Nominal.

**Variables Cuantitativas escogidas:**
- **EDAD:** Cuantitativa Discreta
- **DURACION TOTAL CARRERA (SEMESTRE):** Cuantitativa Discreta
- **VALOR ARANCEL (PESOS):** Cuantitativa Continua (variable continua requerida)

```python
import pandas as pd
import numpy as np

# Base de datos cargada
df = pd.read_excel('07_MATRICULAS_ED_SUPERIOR_METROPOLITANA_2021_ZONA NORTE.xlsx')
df
```

```python

print("Primeras 5 filas")
df.head(5)
```

```python
n = df.shape[0]
print("Tamaño de la muestra:", n)
```

**2. Item 1: Variables Cualitativas:**

**VARIABLE 1: GENERO (Cualitativa Nominal)**

```python

#Cálculo de frecuencia absoluta. La palabra frecuencia representa la frecuencia absoluta.
frecuencia = df.groupby('GENERO').size()

n = df.shape[0]

#Crear el DF con frecuencias absolutas y relativas
tabla = pd.DataFrame({
    'fi': frecuencia,
    'hi': round((frecuencia/n) *100, 1).astype(str) + '%'  # redondear según contexto y valores de los datos.
})
print("--- TABLA DE FRECUENCIA: GENERO ---")
tabla
```

**GRAFICO DE BARRAS PARA GENERO**

```python
import matplotlib.pyplot as plt

#Gráfico de barras

#Considerar que "tabla" es el nombre de la tabla de frecuencias construida

#Variables
categorias = tabla.index
f = tabla['fi']   #Considerar que "Frecuencia" es el tipo de frecuencia a graficar

#Gráfico
fig, ax = plt.subplots(figsize = (6,6))
barras = ax.bar(categorias, f, color = 'r', edgecolor ='black') #Los colores son de libre elección

#Personalización
ax.set_title('Distribucion de Estudiantes por Genero')
ax.set_xlabel('Genero')
ax.set_ylabel('Cantidad de estudiantes (fi)')
ax.bar_label(barras)
plt.xticks(rotation=45) #Rota en 45° las categorías para evitar la superposición
plt.show()
```

**INTERPRETACION DEL GRAFICO:** Observando el grafico muestra la cantidad de Estudiantes por Genero. Existe una mayor presencia de estudiantes de genero femenino, con un total de 80,617 registros, frente a 71,208 estudiantes de genero masculino.

**VARIABLE 2: TIPO DE INSTITUCION (Cualitativa Nominal)**

```python
#Cálculo de frecuencia absoluta. La palabra frecuencia representa la frecuencia absoluta.
frecuencia = df.groupby('TIPO DE INSTITUCION').size()

n = df.shape[0]

#Crear el DF con frecuencias absolutas y relativas
tabla = pd.DataFrame({
    'fi': frecuencia,
    'hi': round((frecuencia/n) *100, 1).astype(str) + '%'  # redondear según contexto y valores de los datos.
})
tabla
```

**GRAFICO CIRCULAR PARA TIPO DE INSTITUCION**

```python
#Gráfico circular

#Considerar que "tabla" es el nombre de la tabla de frecuencias construida
categorias = tabla.index  #Indices de la tabla
h = tabla['fi']   #Considerar que "Frecuencia" es el tipo de frecuencia a graficar

#Gráfico
fig,ax = plt.subplots(figsize = (6,4)) #Tamaño de la figura
circular = ax.pie(h, labels = categorias, autopct= '%.1f%%')

#Personalización
ax.set_title('Título del gráfico')

plt.show()
```

**INTERPRETACION DEL GRAFICO:** Observando el grafico muestra la distribucion de Estudiantes por Tipo de Institucion. Existe un mayor porcentaje de Estudiantes en universidades privadas, con un total de 41,8% de los registros, seguido de las instituciones profesionales con un 37,7%, mientras que en universidades CRUCH abarca el 18,1%, y los centros de formacion tecnica registran el menor porcentaje con apenas un 2,4%.

**VARIABLE 3: AREA CONOCIMIENTO (Cualitativa Nominal)**

```python
#Cálculo de frecuencia absoluta. La palabra frecuencia representa la frecuencia absoluta.
frecuencia = df.groupby('AREA CONOCIMIENTO').size()

n = df.shape[0]

#Crear el DF con frecuencias absolutas y relativas
tabla = pd.DataFrame({
    'fi': frecuencia,
    'hi': round((frecuencia/n) *100, 1).astype(str) + '%'  # redondear según contexto y valores de los datos.
})
print("--- TABLA DE FRECUENCIA: AREA CONOCIMIENTO ---")
tabla
```

**GRAFICO DE BARRAS PARA AREA DE CONOCIMIENTO**

```python
#Gráfico de barras

#Considerar que "tabla" es el nombre de la tabla de frecuencias construida

#Variables
categorias = tabla.index
f = tabla['fi']   #Considerar que "Frecuencia" es el tipo de frecuencia a graficar

#Gráfico
fig, ax = plt.subplots(figsize = (30,6))
barras = ax.bar(categorias, f, color = 'r', edgecolor ='black') #Los colores son de libre elección

#Personalización
ax.set_title('Estudiantes por Area de Conocimiento')
ax.set_xlabel('Area')
ax.set_ylabel('Cantidad de alumnos (fi)')
ax.bar_label(barras)
plt.xticks(rotation=45) #Rota en 45° las categorías para evitar la superposición
plt.show()
```

**INTERPRETACION DEL GRAFICO:** El diagrama refleja la distribucion de la matricula segun la especialidad academica por los alumnos. Se evidencia una clara preferencia por el area de **Administracion** y **Comercio**, el cual lidera con 40,370 de Estudiantes inscritos, seguido por el area de **Tecnologia** con 31,411 y el area de **Salud** con 23,449 reportados. Por lo contrario, en el area de **Ciencias Basicas** registran el menor volumen de Estudiantes, alcanzando apenas 1,172 registros en total.

**2. Item 1: Variables Cuantitativas**

**Variable 4: EDAD (Cuantitativa Discreta)**

```python
#Cálculo de frecuencia absoluta. La palabra frecuencia representa la frecuencia absoluta.
frecuencia = df.groupby('EDAD').size()

n = df.shape[0]

#Crear el DF con todas las frecuencias
tabla = pd.DataFrame({
    'fi': frecuencia,
    'hi': round((frecuencia/n) *100, 1).astype(str) + '%', # redondear según contexto y valores de los datos, mostrando símbolo de '%'
    'Fi':frecuencia.cumsum(),
    'Hi': round((frecuencia/n) * 100,2).cumsum().astype(str) + '%'
})
print("--- TABLA DE FRECUENCIA: EDAD ---")
tabla
```

```python
#Gráfico de barras

#Considerar que "tabla" es el nombre de la tabla de frecuencias construida

#Variables
categorias = tabla.index
f = tabla['fi']   #Considerar que "Frecuencia" es el tipo de frecuencia a graficar

#Gráfico
fig, ax = plt.subplots(figsize = (30,6))
barras = ax.bar(categorias, f, color = 'r', edgecolor ='black') #Los colores son de libre elección

#Personalización
ax.set_title('Distribucion de Estudiantes por Edad')
ax.set_xlabel('Edad (Años)')
ax.set_ylabel('Cantidad de Alumnos (fi)')
ax.bar_label(barras)
plt.xticks(rotation=45) #Rota en 45° las categorías para evitar la superposición
plt.show()
```
